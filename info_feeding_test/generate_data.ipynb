```sage
from sage.all import *

import csv

MAX_SIZE = 6


# Convert a list into a space-separated string
def stringify(values):

    return " ".join(str(x) for x in values)


with open("multi_encoding_data.csv", "w", newline="") as file:

    writer = csv.writer(file)

    writer.writerow([
        "n",
        "one_line",
        "cycle",
        "lehmer",
        "inversion_vector",
        "coxeter_length"
    ])


    for n in range(1, MAX_SIZE + 1):

        for p in Permutations(n):

            one_line = stringify(list(p))

            cycle = str(p.cycle_tuples())

            lehmer = stringify(list(p.to_lehmer_code()))

            inversion = stringify(list(p.to_inversion_vector()))

            length = p.length()

            writer.writerow([
                n,
                one_line,
                cycle,
                lehmer,
                inversion,
                length
            ])


print("Saved multi_encoding_data.csv")
```
