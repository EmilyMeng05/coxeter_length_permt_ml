```sage
from sage.all import *
import csv

# Store the data as a dictionary
data = []

# Generate all permutations with lengths 1 to 6
for n in range(1, 7):

    for p in Permutations(n):

        P, Q = p.robinson_schensted()

        data.append({
            "n": n,
            "permutation": list(p),
            "P": P,
            "Q": Q
        })

with open("rsk_data.csv", "w", newline="") as file:

    writer = csv.writer(file)

    # Header
    writer.writerow([
        "n",
        "permutation",
        "P",
        "Q"
    ])

    # Data
    for row in data:

        writer.writerow([
            row["n"],
            row["permutation"],
            row["P"],
            row["Q"]
        ])

print("Saved rsk_data.csv")
```
